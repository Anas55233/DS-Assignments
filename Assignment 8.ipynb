```python

# Main inventory dictionary containing all product details
shop = {
    'products': {
        'P1': {'name': 'Shoes', 'price': 2500, 'stock': 5},
        'P2': {'name': 'Shirt', 'price': 1500, 'stock': 10},
        'P3': {'name': 'Bag', 'price': 3000, 'stock': 4},
        'P4': {'name': 'Watch', 'price': 4000, 'stock': 3},
        'P5': {'name': 'Headphones', 'price': 2000, 'stock': 6}
    }
}

# Dictionary to store products added to the shopping cart
cart_items = {}


# Display all available products in a Pandas DataFrame
def display_products():
    import pandas as pd

    df = pd.DataFrame.from_dict(
        shop['products'],
        orient='index'
    ).rename_axis('Product ID').reset_index()

    print("\n========== PRODUCTS ==========\n")
    display(df)


# Add products to the shopping cart
def add_cart():
    product_id = input("Enter Product ID: ")

    # Check whether the entered Product ID exists
    if product_id in shop['products']:
        print(f"Product ID {product_id}: {shop['products'][product_id]}\n")

        add_to_cart = int(input("Enter Quantity for add to cart: "))

        # Retrieve current product stock and name
        stock = shop['products'][product_id]['stock']
        product = shop['products'][product_id]['name']

        # Validate the requested quantity
        if add_to_cart <= 0:
            print("Invalid Quantity!")

        # Check stock availability before adding to cart
        elif add_to_cart <= stock:
            # Reduce inventory stock by the requested quantity
            # shop['products'][product_id]['stock'] -= add_to_cart
            stock -= add_to_cart

            # stock = shop['products'][product_id]['stock']

            # Add quantity to cart or update existing quantity
            cart_items[product] = cart_items.get(product, 0) + add_to_cart

            print(f"Add Cart: {product}: {add_to_cart}")
            print(f"Remaining Quantity of {product}: {stock}")

        else:
            print("Insufficient Quantity")

    else:
        print("Not exist")


# Display cart products and calculate the grand total
def view_cart():
    cart = []
    grand_total = 0

    print("\n========== ADD TO CART ==========\n")

    # Iterate through products and quantities in the cart
    for product, qty in cart_items.items():

        # Find matching product details in the inventory
        for detail in shop['products'].values():
            if detail['name'] == product:
                price = detail['price']
                sub_total = qty * price
                grand_total += sub_total

                # Store product details in a list for DataFrame
                cart.append({
                    'Product': product,
                    'Quantity': qty,
                    'Price': price,
                    'Sub_total': sub_total
                })

    # Display cart details
    for product in cart:
        print(
            f"Product: {product['Product']}, "
            f"Quantity: {product['Quantity']}, "
            f"Price: {product['Price']}, "
            f"Sub Total: {product['Sub_total']}"
        )

    # Return cart data and grand total
    return cart, grand_total


# Clear shopping cart and restore inventory stock
def clear_cart():

    # Restore the quantities of cart products to inventory
    for product, qty in cart_items.items():
        for detail in shop['products'].values():
            if detail['name'] == product:
                detail['stock'] += qty

    # Remove all products from the cart
    cart_items.clear()

    print("\nCart cleared successfully!\n")


# Generate checkout bill and apply discount if eligible
def display_checkout():
    import pandas as pd

    # Retrieve cart details and grand total
    cart, grand_total = view_cart()

    # Convert cart data into a Pandas DataFrame
    df = pd.DataFrame(cart)

    print("\n========== CHECKOUT ==========\n")
    display(df)

    # Display the original grand total
    print(f"Grand Total: {grand_total}")

    # Apply 10% discount on purchases of Rs. 5000 or more
    if grand_total >= 5000:
        discount = grand_total - (grand_total * 0.1)
        print("\nAfter 10% Discount:\n")
        print(f"Grand Total: {discount}")

    # Clear cart after checkout
    clear_cart()


# Main program loop
while True:

    print("\n===== Inventory Management System =====")
    print("1. Display Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Display Checkout")
    print("5. Clear Cart")
    print("6. Exit")

    choice = input("\nEnter your choice (1-6): ")

    if choice == "1":
        display_products()

    elif choice == "2":
        add_cart()

    elif choice == "3":
        view_cart()

    elif choice == "4":
        display_checkout()

    elif choice == "5":
        clear_cart()

    elif choice == "6":
        print("Thank you for using Inventory Management System!\n\n")
        break

    else:
        print("Invalid Choice! Please select between 1-6.")
```
