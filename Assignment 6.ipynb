```python
# Ans 1


# Initialize the account balance
balance = 1000

try:
    # Prompt the user to enter a withdrawal amount and convert input to an integer
    user_input = int(input("Enter amount to withdraw: "))

    # Validate that the withdrawal amount is greater than zero
    if user_input <= 0:
        print("Amount must be more than zero.")

    # Check whether the requested withdrawal exceeds the available balance
    elif user_input > balance:
        print("Insufficient balance.")

    else:
        # Calculate the remaining balance after a successful withdrawal
        rem_balance = balance - user_input

        # Display the successfully deducted withdrawal amount
        print(f"Amount Deducted: {user_input}")

        # Display the updated account balance
        print(f"Remaining Bal: {rem_balance}")
            
except ValueError:
    # Handle invalid input when the user enters a non-numeric value
    print("Please enter a valid number.")

finally:
    # Display the closing message regardless of whether the transaction succeeds or fails
    print("Thank you for using the ATM.")
```

```python
# Ans 2


# Initialize the total number of available seats
available_seats = 5

try:
    # Prompt the user to enter the number of tickets and convert input to an integer
    tickets = int(input("How many tickets do you want to book? "))

    # Validate that the requested number of tickets is greater than zero
    if tickets <= 0:
        print("You must book at least 1 ticket.")

    # Check whether the requested tickets exceed the available seats
    elif tickets > available_seats:
        print("Not enough seats available.")

    else:
        # Calculate the remaining seats after a successful booking
        rem_seats = available_seats - tickets

        # Display the booking confirmation message
        print(f"{tickets} ticket(s) booked successfully")

        # Display the number of seats remaining after booking
        print(f"Remaining Seats: {rem_seats}")
            
except ValueError:
    # Handle invalid input when the user enters a non-numeric value
    print("Please enter a valid number.")

finally:
    # Display the closing message regardless of whether the booking succeeds or fails
    print("Thank you for booking with us.")
```

```python
# Ans 10


# Initialize a list containing all available parking slot numbers
available_slots = [101, 102, 103, 104, 105]

# Initialize a list containing parking slots that are already reserved
reserved =  [106, 107, 108, 109, 110]

try:
    # Prompt the user to enter a parking slot number and convert input to an integer
    slot_number = int(input("Enter the parking slot number you want to reserve: "))

    # Check whether the requested parking slot is already reserved
    if slot_number in reserved:
        print(f"Slot {slot_number} is already reserved by someone else.")
        
    # Check whether the requested parking slot exists in the available slots list
    elif slot_number in available_slots:
        print(f"Slot {slot_number} is reserved for you.")

    # Handle cases where the entered parking slot number does not exist in either list
    else:
        print(f"Slot {slot_number} does not exist.")

except ValueError:
    # Handle invalid input when the user enters a non-numeric value
    print("Invalid input. Please enter a valid slot number.")
```
