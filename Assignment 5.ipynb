```python
# Ans 6

# Take input for length and width
length = int(input("Enter Length of Rectangle: "))
width = int(input("Enter Width of Rectangle: "))

# Write function to calculate rectangle area and perimeter by length and width
def calculate_rectangle(length, width):
    def calculate_area():
        area = length * width
        return area
        
    def calculate_perimeter():
        perimeter = 2 * (length + width)
        return perimeter

    # Now return area and perimeter from inner functions
    print(f"\nArea of Rectangle: {calculate_area()} \nPerimeter of Rectangle: {calculate_perimeter()}")

# Now calculate rectangle:
calculate_rectangle(length, width)
```

```python
# Ans 8

# Take input from user to write no. of night and cost perday
number_of_nights = int(input("Enter Number of nights stayed: "))
cost_perday = int(input("Enter Room perday Charges: "))

# Write Function to make total_bill of hotel in which we stayed:
def total_hotel_bill(number_of_nights, cost_perday):
    def room_cost():
        return number_of_nights * cost_perday
        
    # Add function to include tax    
    def cost_tax():
        return room_cost() * (0.1)
        
    # make total bill with tax included
    def total_bill():
        return room_cost() + cost_tax()

    # Now return room cost with tax and makke total bill
    print(f"\nRoom Cost: {room_cost()} \nTax: {cost_tax()} \nTotal Bill: {total_bill()}")

# Call Function total_hotel_bill
total_hotel_bill(number_of_nights, cost_perday)
    
```

```python
# Ans 9

# Take input to write username and password
username = input("Enter Username: ")
password = input("Enter Password: ")

# Make function to check access 
def check_access(username, password):

    # Make inner function to validate username 
    def is_username():
        if " " in username:
            return "Spaces Not Allowed"
            
        elif not username.replace("_", "").isalnum():
            return "Only letters, numbers and underscore are allowed."

        elif len(username) < 8 or len(username) > 20:
            return "username must be atleast 8 characters and maximum 20 characters"
        
        else:
            return username
            
    # Make inner function to validate password 
    def is_password():
        if " " in password:
            return "Spaces Not Allowed"
            
        elif len(password) < 8 or len(password) > 20:
            return "password must be atleast 8 characters and maximum 20 characters"
            
        else:
            return password

# Validate username and password
    if is_username() == username:
        if is_password() == password:
            return "Access Granted"
        else:
            return is_password()
    else:
        return is_username()

# Now print access for login credincials
print(check_access(username, password))
```

```python
# Ans 10

# Take input to write price, qty and discount
price = int(input("Enter Price: "))
qty = int(input("Enter Quantity: "))
discount = int(input("Enter Discount: "))

# Write shopping bill function to check total with or without discount
def shopping_bill(price, qty, discount = 0):
    def calculate_total():
        total = price * qty
        return total

    def apply_discount(total):
        return max(0, total - discount)

    total = calculate_total()
    return apply_discount(total)

print(f"Final Bill = {shopping_bill(price, qty)} PKR (No Discount)")
print(f"Final Bill = {shopping_bill(price, qty, discount)} PKR (With Discount)")
```
