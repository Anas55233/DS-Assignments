```python

# Ans # 1

# Define the Employee class
class Employee:

    # Initialize employee attributes when an object is created
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):

        # Assign the provided values to the object's attributes
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    # Define a method to display complete employee information
    def describe_employee(self):

        # Print all employee attributes in a formatted structure
        print(f"""
Employee Information
--------------------
Name: {self.name}
Employee Number: {self.number}
Address: {self.address}
Salary: Rs. {self.salary}
Post: {self.post}
Year of Joining: {self.year_of_joining}
Education: {self.education}
Year of Experience: {self.year_of_experience}
""")

    # Define a method to update employee's years of experience
    def update_year_of_experience(self, years):

        # Add the given years to the existing experience
        self.year_of_experience += years

        # Return the updated experience
        return f"Updated Experience: {self.year_of_experience} years"

    # Define a method to increase employee's salary
    def increase_salary(self, amount):

        # Add the given amount to the existing salary
        self.salary += amount

        # Return the updated salary
        return f"Updated Salary: Rs. {self.salary}"


# Create the first employee object using the Employee class
employee101 = Employee(
    "Anas Sohail", 101, "Karachi",
    50000, "Python Developer", 2024,
    "Intermediate", 2
)


# Access and print four employee attributes individually
print(employee101.name)
print(employee101.number)
print(employee101.salary)
print(employee101.post)


# Call the method to display employee 101's complete information
employee101.describe_employee()

# Increase employee 101's salary by Rs. 10,000
print(employee101.increase_salary(10000))

# Increase employee 101's experience by 1 year
print(employee101.update_year_of_experience(1))


# Create 9 additional employee objects using the Employee class

# Create employee 102
employee102 = Employee("Ali", 102, "Lahore", 45000, "Data Analyst", 2022, "BS Computer Science", 4)

# Create employee 103
employee103 = Employee("Ahmed", 103, "Karachi", 60000, "AI Engineer", 2023, "BS Software Engineering", 3)

# Create employee 104
employee104 = Employee("Sara", 104, "Islamabad", 55000, "Python Developer", 2021, "BS IT", 5)

# Create employee 105
employee105 = Employee("Hassan", 105, "Multan", 40000, "Web Developer", 2024, "Intermediate", 2)

# Create employee 106
employee106 = Employee("Ayesha", 106, "Lahore", 70000, "Data Scientist", 2020, "MS Data Science", 6)

# Create employee 107
employee107 = Employee("Usman", 107, "Karachi", 50000, "Database Administrator", 2022, "BS Computer Science", 4)

# Create employee 108
employee108 = Employee("Fatima", 108, "Islamabad", 65000, "ML Engineer", 2021, "MS Artificial Intelligence", 5)

# Create employee 109
employee109 = Employee("Bilal", 109, "Peshawar", 35000, "Junior Developer", 2025, "BS IT", 1)

# Create employee 110
employee110 = Employee("Zain", 110, "Quetta", 55000, "Software Engineer", 2022, "BS Software Engineering", 4)


# Store all 10 employee objects in a single list
employees = [
    employee101, employee102, employee103, employee104,
    employee105, employee106, employee107, employee108,
    employee109, employee110
]


# Use a for loop to display information for all 10 employees
for employee in employees:

    # Call describe_employee() for each employee object
    employee.describe_employee()


# Perform salary and experience updates using employee methods

# Increase employee 102's salary by Rs. 5,000
print(f"Employee_102: {employee102.increase_salary(5000)}")

# Increase employee 104's salary by Rs. 10,000
print(f"Employee_104: {employee104.increase_salary(10000)}")

# Increase employee 106's experience by 2 years
print(f"Employee_106: {employee106.update_year_of_experience(2)}")

# Increase employee 108's experience by 1 year
print(f"Employee_108: {employee108.update_year_of_experience(1)}")
```

```python

# Ans # 2

# Function to create employee data and employee-related operations
def employee_data(name, number, address, salary, post,
                 joining_year, education, year_of_experience):

    # Store employee details in a dictionary
    employee = {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "joining_year": joining_year,
        "education": education,
        "experience": year_of_experience
    }

    # Nested function to display complete employee information
    def employee_info():
        return f"""
Employee Information
--------------------
Name: {employee["name"]}
Employee Number: {employee["number"]}
Address: {employee["address"]}
Salary: Rs. {employee["salary"]}
Post: {employee["post"]}
Year of Joining: {employee["joining_year"]}
Education: {employee["education"]}
Year of Experience: {employee["experience"]}
"""

    # Nested function to update employee's years of experience
    def update_year_of_experience(years):
        employee["experience"] += years
        return f"Updated Experience: {employee['experience']} years"

    # Nested function to increase employee's salary
    def increase_salaries(amount):
        employee["salary"] += amount
        return f"Updated Salary: Rs. {employee['salary']}"

    # Nested function to update employee's job post
    def updated_post(post):
        employee['post'] = post
        return f"Updated Post: {employee['post']}"

    # Return employee dictionary and nested functions as a dictionary
    return {
        "employee": employee,
        "info": employee_info,
        "salary": increase_salaries,
        "experience": update_year_of_experience,
        "update_post": updated_post
    }


# Create employee 101 and store its returned dictionary
employee_101 = employee_data("Anas Sohail", 101, "Karachi", 50000, "Python Developer", 2024, "Intermediate", 2)

# Create employee 102
employee_102 = employee_data("Ali", 102, "Lahore", 45000, "Data Analyst", 2022, "BS Computer Science", 4)

# Create employee 103
employee_103 = employee_data("Ahmed", 103, "Karachi", 60000, "AI Engineer", 2023, "BS Software Engineering", 3)

# Create employee 104
employee_104 = employee_data("Sara", 104, "Islamabad", 55000, "Python Developer", 2021, "BS IT", 5)

# Create employee 105
employee_105 = employee_data("Hassan", 105, "Multan", 40000, "Web Developer", 2024, "Intermediate", 2)

# Create employee 106
employee_106 = employee_data("Ayesha", 106, "Lahore", 70000, "Data Scientist", 2020, "MS Data Science", 6)

# Create employee 107
employee_107 = employee_data("Usman", 107, "Karachi", 50000, "Database Administrator", 2022, "BS Computer Science", 4)

# Create employee 108
employee_108 = employee_data("Fatima", 108, "Islamabad", 65000, "ML Engineer", 2021, "MS Artificial Intelligence", 5)

# Create employee 109
employee_109 = employee_data("Bilal", 109, "Peshawar", 35000, "Junior Developer", 2025, "BS IT", 1)

# Create employee 110
employee_110 = employee_data("Zain", 110, "Quetta", 55000, "Software Engineer", 2022, "BS Software Engineering", 4)


# Print individual attributes of employee 101
print(employee_101["employee"]["name"])
print(employee_101["employee"]["number"])
print(employee_101["employee"]["salary"])
print(employee_101["employee"]["post"])


# Store all 10 employee dictionaries in a single list
employeess = [
    employee_101, employee_102, employee_103, employee_104,
    employee_105, employee_106, employee_107, employee_108,
    employee_109, employee_110
]


# Display complete information of all employees using a loop
for employee in employeess:
    print(employee["info"]())


# Perform updates using the nested functions

# Increase employee 102's salary by Rs. 5,000
print(f"Employee_102: {employee_102['salary'](5000)}")

# Increase employee 104's salary by Rs. 10,000
print(f"Employee_104: {employee_104['salary'](10000)}")

# Increase employee 106's experience by 2 years
print(f"Employee_106: {employee_106['experience'](2)}")

# Increase employee 108's experience by 1 year
print(f"Employee_108: {employee_108['experience'](1)}")

# Update employee 108's post
print(f"Employee_108: {employee_108['update_post']('BS IT')}")


# Display updated information of all 10 employees
for employee in employeess:
    print(employee["info"]())
```
