```python
questions = [
    "How was the food quality?",
    "How was the customer service?",
    "How clean was the restaurant?",
    "How was the price of the food?",
    "How was your overall experience?"
]


rating = []
valid_rating = True

for question in questions:
    print(question)
    print("\n===== Rating =====")
    print("1. Poor")
    print("2. Average")
    print("3. Good")
    print("4. Very Good")
    print("5. Excellent")
    try:
        rate_number = int(input("Enter Raing Number:"))
        
        if rate_number >= 1 and rate_number <= 5:
            print(f"Rating: {rate_number}")
        else:
            print("Invalid Choice! Please select between 1-5.")
            valid_rating = False
            break

    except ValueError:
        print("Please enter a valid digit!")
        valid_rating = False
        break
    
    print("\n\n")
    rating.append(rate_number)

if valid_rating:
    
    print("Rating", rating)
    total_rate = sum(rating)
    print(f"\n Total rating: {total_rate}")
    avg_rate = total_rate/len(questions)
    print(f"Average Rating: {avg_rate}\n")
    
    print("Over All:\n")
    if avg_rate >= 4.0:
        print("Excellent Experience")
    
    elif avg_rate >= 3.0 and avg_rate < 4.0:
        print("Good Experience")
    
    elif avg_rate >= 2.0 and avg_rate < 3.0:
        print("Average Experience")
    
    else:
        print("Needs Improvement")

```
